```python

# First Order Logic - Forward Chaining

# Initial Knowledge Base (Facts)
KB = [
    ("American", "Robert"),
    ("Missile", "T1"),
    ("Owns", "A", "T1"),
    ("Enemy", "A", "America")
]

# Rules
rules = [
    ([("Missile", "?x")],
     ("Weapon", "?x")),

    ([("Missile", "?x"), ("Owns", "A", "?x")],
     ("Sells", "Robert", "?x", "A")),

    ([("Enemy", "?x", "America")],
     ("Hostile", "?x")),

    ([("American", "?p"), ("Weapon", "?q"),
      ("Sells", "?p", "?q", "?r"),
      ("Hostile", "?r")],
     ("Criminal", "?p"))
]

# Function to match facts with rules
def match(pattern, fact, substitution):
    if pattern[0] != fact[0]:
        return None

    sub = substitution.copy()

    for p, f in zip(pattern[1:], fact[1:]):
        if p.startswith("?"):
            if p in sub and sub[p] != f:
                return None
            sub[p] = f
        elif p != f:
            return None

    return sub


# Function for forward chaining
def forward_chaining(KB, rules, query):

    while True:
        new_facts = []

        for premises, conclusion in rules:

            substitutions = [{}]

            for premise in premises:
                updated = []

                for sub in substitutions:
                    for fact in KB:
                        result = match(premise, fact, sub)

                        if result is not None:
                            updated.append(result)

                substitutions = updated

            for sub in substitutions:
                new_fact = tuple(
                    sub.get(x, x) if x.startswith("?") else x
                    for x in conclusion
                )

                if new_fact not in KB and new_fact not in new_facts:
                    new_facts.append(new_fact)

        if not new_facts:
            break

        KB.extend(new_facts)

        print("New Facts:", new_facts)

        if query in KB:
            return True

    return query in KB


# Query to prove
query = ("Criminal", "Robert")

# Execute
if forward_chaining(KB, rules, query):
    print("\nGoal Proved: Robert is Criminal")
else:
    print("\nGoal Not Proved")
```
